# Pengantar Web Mining

**Web Mining** merupakan proses pemanfaatan metode *data mining* untuk mengambil dan mengolah informasi yang terdapat di internet. Melalui proses ini, data dari berbagai halaman web dapat dianalisis sehingga menghasilkan pola atau informasi yang dapat digunakan untuk kebutuhan tertentu.

Data yang terdapat pada web memiliki bentuk yang beragam, mulai dari teks, gambar, tautan, hingga catatan aktivitas pengguna. Berdasarkan jenis data dan cara pengolahannya, Web Mining secara umum terbagi menjadi tiga bagian, yaitu:

### 1. Web Content Mining

Web Content Mining berfokus pada pengambilan informasi yang terdapat di dalam suatu halaman web. Data yang dapat dianalisis meliputi teks, gambar, video, audio, maupun informasi pendukung lainnya.

**Contoh:** mengambil artikel berita dari website untuk kemudian dilakukan klasifikasi atau analisis terhadap isi beritanya.

### 2. Web Structure Mining

Web Structure Mining digunakan untuk mempelajari hubungan antara satu halaman web dengan halaman web lainnya melalui struktur *hyperlink*. Dengan pendekatan ini, hubungan antarhalaman dapat dianalisis seperti sebuah jaringan atau graf.

**Contoh:** penggunaan konsep *PageRank* untuk melihat tingkat kepentingan sebuah halaman berdasarkan hubungan tautan yang dimilikinya.

### 3. Web Usage Mining

Web Usage Mining berhubungan dengan pengolahan data mengenai aktivitas pengguna ketika menggunakan sebuah website. Informasi tersebut dapat membantu mengetahui pola dan kebiasaan pengguna selama berinteraksi dengan suatu situs.

**Contoh:** menganalisis riwayat kunjungan dan aktivitas pengguna untuk memberikan rekomendasi yang lebih sesuai.

### Kesimpulan

Ketiga jenis Web Mining memiliki fokus yang berbeda. **Web Content Mining** digunakan untuk memahami informasi yang terdapat dalam halaman web, **Web Structure Mining** digunakan untuk menganalisis hubungan antarhalaman, sedangkan **Web Usage Mining** digunakan untuk mempelajari aktivitas pengguna ketika berinteraksi dengan website.
